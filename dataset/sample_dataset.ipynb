````python
import json
import random
from pathlib import Path

DATASET_PATH = Path("conflict_dataset.jsonl")
N_SAMPLE = 10

def load_dataset(path):
    rows = []
    with open(path, "r", encoding="utf-8") as f:
        for line in f:
            line = line.strip()
            if not line:
                continue
            rows.append(json.loads(line))
    return rows

dataset = load_dataset(DATASET_PATH)
samples = random.sample(dataset, min(N_SAMPLE, len(dataset)))
print(f"Loaded {len(dataset)} examples, sampled {len(samples)}.")
````

````python
import html
from IPython.display import display, Markdown, HTML

def show_example(item, index):
    sys_p = item['system_prompt'].replace('```', '"')
    task = item['task_prompt']
    conflict = item['conflict_instruction']
    task_esc = html.escape(task)
    conflict_esc = html.escape(conflict)
    if item['conflict_position'] == 'prefix':
        user_part = f'<span style="color:red">{conflict_esc + " "}</span>{task_esc}'
    else:
        user_part = f'{task_esc}<span style="color:red">{" " + conflict_esc}</span>'

    md = f"""---
**{item['constraint_id']}** ({item['verification_class']})
`sample_id`: `{item['sample_id']}` · conflict position: **{item['conflict_position']}**

**System prompt** (model instructed to follow):
```
{sys_p}
```

**User prompt** (conflict in red, no space before task):
"""
    display(Markdown(md))
    display(HTML(f"<p style=\"white-space:pre-wrap;\">{user_part}</p>"))
    display(Markdown(f"`wildchat_id`: `{item['wildchat_id']}` · `instruction_args`: `{item['instruction_args']}`"))

for i, item in enumerate(samples):
    show_example(item, i)
````
